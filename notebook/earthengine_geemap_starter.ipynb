# Earth Engine e geemap

Notebook inicial para testar autenticação do Google Earth Engine e visualizar um mapa simples com geemap.

```python
import ee
import geemap

print('Imports carregados com sucesso')
```

```python
# Inicialização robusta do Earth Engine:
# - tenta usar GOOGLE_CLOUD_PROJECT se definida
# - se não houver, usa o project ID 'sugarcanemle' como padrão
import os
project = os.environ.get('GOOGLE_CLOUD_PROJECT') or os.environ.get('EE_PROJECT') or 'sugarcanemle'
try:
    if project:
        ee.Initialize(project=project)
    else:
        ee.Initialize()
    print(f'Earth Engine inicializado (project={project})')
except ee.EEException as e:
    msg = str(e)
    if 'no project found' in msg or 'NO_PROJECT_EXCEPTION' in msg:
        raise ee.EEException(
            "ee.Initialize: nenhum projeto encontrado. Defina a variável de ambiente GOOGLE_CLOUD_PROJECT ou passe `project='SEU_PROJECT_ID'`.\n"
            "No terminal:\n\n  gcloud auth application-default login\n  gcloud config set project SEU_PROJECT_ID\n\n"
            "Ou no notebook:\n  import os\n  os.environ['GOOGLE_CLOUD_PROJECT']='SEU_PROJECT_ID'\n  ee.Initialize(project='SEU_PROJECT_ID')"
        ) from e
    else:
        raise
```

```python
# Mapa simples centrado em Ribeirão Preto
import ipywidgets as widgets
m = geemap.Map(center=[-21.1775, -47.8103], zoom=9)
popup = widgets.HTML('Ribeirão Preto')
m.add_marker(location=[-21.1775, -47.8103], popup=popup)
m
```
